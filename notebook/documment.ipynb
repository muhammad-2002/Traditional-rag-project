```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

def split_document(pdf_documents):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
        separators=["\n\n", "\n", " ", ""]
    )

    split_doc = text_splitter.split_documents(pdf_documents)

    print(f"Original documents: {len(pdf_documents)}")
    print(f"Total chunks: {len(split_doc)}")

    if split_doc:
        print(f'\nExample chunk:')
        print(f'Content:{split_doc[0].page_content}')
        print(f'MetaData:{split_doc[0].metadata}')

    return split_doc
```

#### Embadings

```python
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings
import uuid
from typing import List,Dict,Any,Tuple
from sklearn.metrics.pairwise import cosine_similarity
```

```python
class EmbeddingManager:
    def __init__(self,model_name:str='all-MiniLM-L6-V2'):
        self.modle_name =model_name
        self.model =None
        self._load_model()

    def _load_model(self):
        try:
            print(f"Loading Embedding model {self.modle_name}")
            self.model = SentenceTransformer(self.modle_name)
            print(f"Model loaded successfully. Embedding dimension: {self.model.get_sentence_embedding_dimension()}")
        except Exception as e:
            print(f"Error loading model {self.modle_name}:{e}")
            raise
    def generate_embeddings(self,texts:List[str])->np.ndarray:
        if not self.model:
            raise ValueError("model not loaded")
        print(f'generate embedding for {len(texts)} strings..')
        embeddings =self.model.encode(texts,show_progress_bar=True)
        print(f'generate Embedding with shape{embeddings.shape}')
        return embeddings
```

```python
# initialize object
embedding_manager =EmbeddingManager()
embedding_manager
```

### vector_store

```python
class VectorStore:
    def __init__(self,collection_name:str='pdf_documents',persist_directory:str='../data/vector_store'):
        self.collection_name =collection_name
        self.persist_directory =persist_directory
        self.client =None
        self.collection =None
        self._initialize_store()

    def _initialize_store(self):
        try:
            # persistent chromadb client
         os.makedirs(self.persist_directory,exist_ok=True)
         self.client =chromadb.PersistentClient(path=self.persist_directory)
         #create collection
         self.collection =self.client.get_or_create_collection(
             name=self.collection_name,
             metadata={"description":"pdf documment embeddings for RAG"}
         )
         print(f"Vector store initialized: {self.collection_name}")
         print(f'Existing documment in collection:{self.collection.count()}')
        except Exception as e:
           print(f'Error initializing vector store {e}')

    def add_documments(self,documments:List[Any],embeddings:np.ndarray):
       if len(documments)!=len(embeddings):
          raise ValueError("Number documments must match number of embeddings")

       print(f"Adding { len(documments)} to vectore store...")

       #    prepair data for cromadb
       ids=[]
       metadatas=[]
       documments_text=[]
       embeddings_list =[]
       

       for i,(doc,embedding) in enumerate(zip(documments,embeddings)):
        doc_id = f"doc_{uuid.uuid4().hex[:8]}_{i}"
        ids.append(doc_id)
        metadata =dict(doc.metadata)
        metadata['index'] =i
        metadata['content_length']=len(doc.page_content)

        metadatas.append(metadata)
        documments_text.append(doc.page_content)
        embeddings_list.append(embedding)
       try:
         self.collection.add(
            ids=ids,
            embeddings=embeddings_list,
            metadatas=metadatas,
            documents =documments_text
         )
         print(f'successfully added {len(documments)} to vector store')
         print(f'Total documment in collection {self.collection.count()}')
       except Exception as e:
         print(f"Error adding documment to vector store: {e}")
```

#### Retriever pipeline from vector store

```python
class RAGRetriever:
    def __init__(self,vectorstore:VectorStore,embedding_manager:EmbeddingManager):
        self.vectore_store =vectorstore
        self.embedding_manager =embedding_manager
    def retrieve(self,querry:str,top_k:int=5,score_thrsehold:float=0.0)->List[dict[str,Any]]:
        print(f"Retrieve Documment for {querry}") 
        print(f'Top_k {top_k} score thrsehold {score_thrsehold}')
        querry_embedding =self.embedding_manager.generate_embeddings(querry)
        # search in vector store
     
        try:
            result =self.vectore_store.collection.query(
                query_embeddings=querry_embedding.tolist(),
                n_results=top_k
            )
            # process result
            
            retrieve_doc=[]
            if result['documents'] and result['documents'][0]:
                documments=result['documents'][0]
                metadatas=result['metadatas'][0]
                distances =result['distances'][0]
                ids=result['ids'][0]
                for i,(doc,meta,dist,id) in enumerate(zip(documments,metadatas,distances,ids)):
                    similarity_score =1-dist
                    if similarity_score>=score_thrsehold:
                        retrieve_doc.append({
                              "id":id,
                              "content":doc,
                              "metadata":meta,
                              "similarity_score":similarity_score,
                              'distance':dist,
                              "rank":i+1
                          })
                print(f'Retrieve {len(retrieve_doc)} after filltering..')
            else:
                print('No documment Found...')
            return retrieve_doc
        except Exception as e:
            print(f'Error during retrieve {e}')
            return []
```

```python
rag_retriever=RAGRetriever(vectorstore,embedding_manager)
rag_retriever.retrieve("what is Inventory Management System")
```

#### Integration vectordb context pipeline with LLM output

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
groq_api_key =os.getenv('GROQ_API_KEY')

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=groq_api_key,
    temperature=0
)
# simple RAG function :Retrieve context + generate Response
def simple_rag(query,retrieval,llm,top_k:int=3):
    results = retrieval.retrieve(query,top_k=top_k)
    context ='\n\n'.join(doc['content'] for doc in results) if results else ""
  
   
    if not content:
        print("No result fount to relevent context")
    #generate answer using groq llm
    prompt =f"""Use the following context to answer qustion concisely.
           Context:{context}
           Query:  {query}
           Answer:
    """
    
    response_llm =llm.invoke([prompt.format(content=context,query=query)])
    return response_llm.content
```

```python
answer =simple_rag("what is Software Requirements Specification (SRS)?",rag_retriever,llm=llm)
print(answer)
```
